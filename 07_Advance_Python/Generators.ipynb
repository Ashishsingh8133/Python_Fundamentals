## Generator Functions

Using `yield` to produce values one at a time instead of returning the complete result at once.


```python
def square(n):
    for i in range(3):
        yield i**2  #this yield keyword makes the square function a generator
```

```python
square(3)
```

```python
for i in square(3): #here as we can see that square(3) now a generator for everytime 'i' will go to square function and generate the square of next element and it will print it
  print(i)
```

```python
def square(n):
    for i in range(3):
        yield i**2
```

```python
a=square(4)
```

## Using `next()`

Retrieving the next value from a generator.


```python
try:
    print(next(a))
except StopIteration:
    print("there are no elements")
```

```python
def square(n):
    for i in range(4):
        yield i**2
```

```python
##read the large file using generator
def large_file_read(file_path):
    with open('example.txt',mode='r') as file:
        for line in file:
            yield line
```

```python
file_path='example.txt'

for line in large_file_read(file_path):
    print(line.strip())
```
