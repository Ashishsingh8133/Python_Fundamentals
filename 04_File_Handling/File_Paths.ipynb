# File Paths and Directories

Managing file paths and directories using Python’s `os` module.

## Current Working Directory

Uses `os.getcwd()` to find the directory from which Python is currently running.

```python
# Current Working Directory

import os
cwd=os.getcwd()
print(f"current working directory is: {cwd}")
```

## Creating a Directory

Uses `os.mkdir()` to create a new directory.

```python
#create directory 
import os
new_directory='package'
os.mkdir(new_directory)
print(f"Directory {new_directory} is created")
```

## Listing Files and Directories

Uses `os.listdir()` to return items inside the specified directory.

```python
#listing all the files and directories
items=os.listdir('.')
print(items)
```

## Joining Paths

Uses `os.path.join()` to build a platform-independent file path.

```python
#joining paths

dir_name='folder'
file_name='ashish.txt'
full_path=os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

## Checking Whether a Path Exists

Uses `os.path.exists()` to test whether the specified path exists.

```python
# Checking Whether a Path Exists

path='example.txt'
if os.path.exists(path):
    print(f"the path {path} is existed")
else:
    print(f"the path {path} does not existed")
```

## Checking File or Directory Type

Uses `os.path.isfile()` and `os.path.isdir()` to determine the type of a path.

```python
## check whether the given path is file or directory or neither file nor directory

path='pac'
if os.path.isfile(path):
    print(f"the given {path} is a file")
elif os.path.isdir(path):
    print(f"the given {path} is directory")
else:
    print(f"the given {path} is nor file neither a directory")
```

## Key Takeaways

- Use context managers (`with open(...)`) to safely work with files.
- Choose the file mode (`r`, `w`, `a`, `rb`, `wb`, `w+`) based on the required operation.
- Use `os` functions to inspect paths, directories, and the current working directory.
