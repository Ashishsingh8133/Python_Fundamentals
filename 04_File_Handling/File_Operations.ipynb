# File Operations

Working with text and binary files using Python file handling methods.

## Reading File Content

Uses `open()` with read mode (`r`) and `read()` to load the complete file content.

```python
## Reading a file content
with open('example.txt',mode='r') as file:
    content=file.read()
    print(content)
```

## Reading a File Line by Line

Iterates directly over the file object so each line can be processed separately.

```python
#Reading file line by line

with open('example.txt',mode='r') as file:
    for line in file:
        print(f"{line}")
```

## Removing Newline Characters

Uses `strip()` to remove surrounding whitespace, including the newline character.

```python
# Removing Newline Characters

with open('example.txt',mode='r') as file:
    for line in file:
        print(line.strip()) ## sstrip() removes the newline character
```

## Writing to a File — Overwrite Mode

`w` creates or overwrites the file, so existing content is replaced.

```python
## write a file (overwriting)

with open("example.txt",mode='w') as file:
    file.write('hi am krish i am good\n')

## the problem is with this approach is when we directly write anything in file all the old content got deleted
```

## Appending to a File

`a` adds new content to the end of the existing file without removing previous content.

```python
## write a file (without overwritten)
with open('example.txt',mode='a') as file:
    file.write('append mode is taking place\n')

    ##by using this approach all the old content will remain stable and append the new content
```

## Writing Multiple Lines

`writelines()` writes a sequence of strings to the file.

```python
## write multiple lines in file at once
lines=['hi ashish\n','are you in UK\n','i hope everything is going well there\n']
with open('example.txt',mode='a')as file:
    content=file.writelines(lines)
```

## Writing a Binary File

`wb` opens a file for binary writing. The `b` prefix creates bytes data.

```python
##write into binary files
data=b'\x00\x01\x02\x03'
with open('example.bin',mode='wb') as file:
    file.write(data)
```

## Reading a Binary File

`rb` opens a file for binary reading and returns bytes.

```python
##read binary file

with open('example.bin',mode='rb') as file:
    content=file.read()
    print(content)
```

## Copying Content Between Files

Reads content from a source file and writes it into a destination file.

```python
# read the content from one file and write it into another file 

with open('example.txt',mode='r') as source_file:
    content=source_file.read()

with open('dest.txt',mode='w') as dest_file:
    dest_file.write(content)
    
```

## Counting Lines, Words, and Characters

Defines a reusable function using `readlines()`, `split()`, `len()`, and `sum()`.

```python
#count the number of lines,words and chracters in a file

def count_text_files(file_path):
    with open(file_path,mode='r') as file:
        lines=file.readlines()
        line_count=len(lines)
        word_count=sum(len(line.split()) for line in lines)
        char_count=sum(len(line) for line in lines)
    return line_count,word_count,char_count 

file_path='example.txt'
line,word,char=count_text_files(file_path)
print(f"number of lines in files is: {line}")
print(f"number of words are in file:{word}")
print(f"number of char in file is:{char}")
```

## Writing and Reading with `w+`

`w+` allows both writing and reading. `seek(0)` moves the cursor back to the beginning before reading.

```python
#write the file and read it using 'w+' mode

with open('example.txt',mode='w+') as file:
    file.write('hello this is new concet\n')
    file.write('we need to learn this\n')

    #for reading we have to move the cusror to the beginning of the file
    file.seek(0)

    content=file.read()
    print(content)
```

## Reading Binary Data

Reads the binary file using `rb` mode.

```python
# Reading Binary Data

with open('example.bin',mode='rb') as file:
    content=file.read()
    print(content)
```

## Creating Binary Data

Creates bytes and writes them to a binary file using `wb` mode.

```python
# Creating Binary Data

data=b'\x00\x01\x02\x03'
with open('tyu.bin',mode='wb') as file:
    file.write(data)
    
```

## Key Takeaways

- Use context managers (`with open(...)`) to safely work with files.
- Choose the file mode (`r`, `w`, `a`, `rb`, `wb`, `w+`) based on the required operation.
- Use `os` functions to inspect paths, directories, and the current working directory.
