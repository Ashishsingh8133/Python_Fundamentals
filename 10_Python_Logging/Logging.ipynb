## Configuring Logging

Using `logging.basicConfig()` to configure the logging system.


```python
import logging

##configure the basic logging settings 
logging.basicConfig(level=logging.DEBUG)

###log messages with different severity levels
logging.debug("this is debug message")
logging.info("this is info message")
logging.warning("this is warning message")
logging.error("this is error message")
logging.critical("This is a critical message")
```

```python
##basic configuration of logging
import logging
logging.basicConfig(
    filename='app.log', ##this is a log file which will save all the logs inside it 
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s', ##format
    datefmt='%y-%m-%d %H:%M:%S' ##date format and time format
)

###log messages with different severity levels
logging.debug("this is debug message")
logging.info("this is info message")
logging.warning("this is warning message")
logging.error("this is error message")
logging.critical("This is a critical message")

logging.debug("this is debug message")
logging.info("this is info message")
logging.warning("this is warning message")
logging.error("this is error message")
logging.critical("This is a critical message")
```
