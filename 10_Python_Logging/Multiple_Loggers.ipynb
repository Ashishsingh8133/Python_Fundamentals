## Working with Multiple Loggers

Using separate loggers to organise logging for different parts of an application.


```python
#create logger 1 for module 1
import logging


logger1=logging.getLogger("module1")
logger1.setLevel(logging.DEBUG)


logger2=logging.getLogger("module2")
logger2.setLevel(logging.WARNING)


logging.basicConfig(
level=logging.DEBUG,
format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
datefmt='%Y-%M-%D %H:%M:%S'
)

logger1.debug("this is debug case for logger1")
logger2.warning("this is warning case for logger2")
```
