```python

##Shallow copy 
##in this when we copy one dictionary into another directly then after when we update key value in the original dictionary
##that change also reflect into the dictionary in which we copied the original one 
student={ "name":"ashish","age":67,"marks":89}
student_copy=student
```

```python
print(student_copy)
```

```python
##here you can see that we are updating the value of key name that change will also reflect in copied dictionary
student["name"]="nirmal"
print(student)
```

```python
##here you can clearly see the reflection name has changed in both dictionary
print(student_copy)
```

```python
#i am (.copy()) because when we copy one dictionary into another dictionary then
#after that when we update any value in the original dictionary that value also changed in the another copied dictionary which should not happen for which we use .copy() to prevent this issue
student={"name":"ashish","age":28,"marks":89,"religion":"hindu"}
student_copy=student #copy without using copy
print(student_copy)
```

```python
student["Surname"]="Singh" #changing value in student
print(student) 
```

```python
print(student_copy) # as u can see that changed value is also reflecting here
```

```python
#solution of shallow copy method 
student={"name":"ashish","age":28,"marks":89,"religion":"hindu"}
student_copy1=student.copy() #using this method we can prevent the above issue
print(student_copy1)
```

```python
student["name"]="Nirmal" #update value 
```

```python
#nested dictionaries 
students={
    "student1":{"name":"ashish","age":22,"marks":78},
    "student2":{"name":"anmol","age":28,"marks":89},
    "student3":{"name":"abhay","age":27,"marks":91}
}
print(students)
```

```python
#accessing values in nested dictionaries
print(students["student1"]["name"])
print(students["student1"]["age"])
```

```python
#iterating over nested dictionaries
students={
    "student1":{"name":"ashish","age":22,"marks":78},
    "student2":{"name":"anmol","age":28,"marks":89},
    "student3":{"name":"abhay","age":27,"marks":91}
}

for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")

```

```python
##dictionary comprehension
lst_dict={x:x**2 for x in range(5)}
print(lst_dict)
```

```python
evennumbers_dictionary={a:a**2 for a in range(10) if a%2==0}
print(evennumbers_dictionary)
```

```python
#count Frequency of elements in list  using dictionary 
numbers=[1,2,2,3,3,3,4,4,4,5,5,5] #list created
frequency={} #dictionary created

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```
