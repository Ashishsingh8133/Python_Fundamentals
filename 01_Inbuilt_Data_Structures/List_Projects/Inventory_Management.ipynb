```python
#Manage an inventory
inventory=['apple','banana','orange','cherry']

#add an item in inventory
inventory.append("grapes")

#remove the item which is out of stock
inventory.remove("banana")

#check whether the item is present in inventory or not
item="cherry"
if item in inventory:
    print(f"currently {item} is in stock")
else:
    print(f"currently {item} is out of stock")


item1="banana"
if item1 in inventory:
    print(f"currently {item1} is in stock")
else:
    print(f"currently {item1} is out of stock")

print("inventory:")
for items in inventory:
    print(f"-{items}")

```
