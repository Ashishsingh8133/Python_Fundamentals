```python
##To-do-List

to_do_list=["clean house","pay bills","study"]

##add tasks into to do list 
to_do_list.append("clean backyard")
to_do_list.append("call to mom")

#remove completed tasks
to_do_list.remove("clean house")
if "clean house" not in to_do_list:
    print("house is cleaned now")

if "pay bills" in to_do_list:
    print("do not forget to pay utility bills")
print("remaining to do list")

for tasks in to_do_list:
    print(f"-{tasks}")
```
