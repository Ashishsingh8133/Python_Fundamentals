```python
empty_tuple=()
print(empty_tuple, type(empty_tuple))
```

```python
##we can create a list and tuple using list and tuple class
lst=list()
print(lst,type(lst))
tpl=tuple()
print(tpl,type(tpl))
```

```python
numbers=tuple(list([1,2,3,4,5,6,7,8,9])) #type casting list into tuple
print(numbers)
```

```python
tpl=tuple((4,5,6,7,8,9,10))
print(tpl)
```

```python
##Accessing tuple elements
print(tpl[0])
print(tpl[-1])
print(tpl[len(tpl)//2]) #used for printing middle element
```

```python
#slicing is Tuple
print(tpl[0:4])
print(tpl[::])
print(tpl[::-2])
```

```python
#mixed tuples
numbers=tuple(list([1,2,3,4,5,6,7,8,9])) #type casting list into tuple
print(numbers)
mixed_tuple=tuple([1,'ashish',9.0,True])
print(mixed_tuple)
```

```python
#Tuples operations 
print(mixed_tuple)
print(numbers)

#concatenate 
concatenated_tuple= mixed_tuple + numbers
print(concatenated_tuple)
```

```python
#multiply
mixed_tuple * 3
```

```python
## tuples are immutable can't change the assigned values

numbers=tuple([1,2,3,4,5])
numbers
numbers[1]='ashish' ## as u can see that data elements assigned to tuples can't change it's immutable nature of tuple
```

```python
#tuple methods
number=tuple([5,5,6,6,7,8,9,10])
print(number.count(5))
print(number.index(8))
```

```python
##packing and unpacking tuples
packed_tuple=1,"ashish",3.14
print(packed_tuple)
```

```python
#Unpacking the tuples
packed_tuple=1,'ashish',7.89
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
#unpack using * 
packed_tuple=(1,2,3,4,5,6,7)
first,*middle,last=packed_tuple
print(first)
print(*middle)
print(last)
```
