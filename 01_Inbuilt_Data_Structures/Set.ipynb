```python
#creating a empty set
my_set={1}
print(type(my_set))
```

```python
empty_set=set((1,2,3,4,5,6))
print(empty_set)
```

```python
#set operations
#add remove elements
my_set={1,2,3,4,5,6,7}
print(my_set)
my_set.add(9) #add
print(my_set)
my_set.add(5) # here i m adding 5 which becomes the duplicate element here set will just ignore it will not add it
print(my_set)
my_set.remove(4) #remove 
print(my_set)
```

```python
#removing the element which is not present in set
print(my_set)
my_set.remove(11) #remove the element in the set which is not present in set by using remove method it will give an error
print(my_set)
```

```python
## to avoid error for such elements we will use Discard()
print(my_set)
my_set.discard(11) ## see no error even though the element we are removing is not in set
print(my_set)
```

```python
#pop method 
my_set={4,5,6,7,8,9,10}
removed_element=my_set.pop()
print(removed_element)
print(my_set)
```

```python
## membership function in set
#it is used for when we want to know whether a data element is a member of set or not means it is present in set or not
my_set={7,8,9,10,11,23}
print(10 in my_set)
print(24 in my_set)
```

```python
#set operations
#union,intersection,intersection_update
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

union_set=set1.union(set2)
print(union_set)

intersection_set=set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
#difference in set
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

difference_set=set1.difference(set2) #data elements that are present in set1 but not in set set2
print(difference_set)

difference_set1=set2.difference(set1) #data elements that are present in set2 but not in set1
print(difference_set1)
```

```python
#symmetric difference in set
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

sym_diff=set1.symmetric_difference(set2) #it will remove all the common elements from both the sets like 4,5,6 and print the combined difference
print(sym_diff)
```

```python

#some other set methods 
set1={1,2,3,4,5,6,7}
set2={1,2,3,4,5,6}

#subset
print(set1.issubset(set2))

#superset
print(set1.issuperset(set2))
```

```python
list=[1,2,3,2,4,5,4,6,7,6,6]
print(list)
unique_set=set(list)
print(unique_set)
```

```python
#counting unique words in text
text="hi, i am ashish and i am  in love with suman and it's very hard for me to live miles away from her, but these distances will never lessen my love for her"
words=text.split()

#convert list of words into a unique set
unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
