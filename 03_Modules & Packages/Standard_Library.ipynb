# Python Standard Library

A concise overview of useful **Python Standard Library** modules for common programming tasks.

**Covered:** `array`, `math`, `random`, `os`, `shutil`, `json`, `csv`, `datetime`, `time`, and `re`.


## 1. Array
`array` provides a typed sequence that can be more memory-efficient than a regular list for numeric data.

```
import array

numbers = array.array("i", [1, 2, 3, 4, 5])
print(numbers)
```

## 2. Math
The `math` module provides common mathematical functions and constants.

```
import math

print(math.sqrt(16))
print(math.pi)
```

## 3. Random
The `random` module is useful for generating random values and selecting random items.

```
import random

print(random.randint(1, 10))  # Random integer from 1 to 10
print(random.choice(["apple", "banana", "cherry"]))  # Random item
```

## 4. OS
The `os` module provides functions for interacting with the operating system and file system.

```
import os

print(os.getcwd())  # Current working directory
```

## 5. Shutil
`shutil` provides high-level file and directory operations. The example below copies a temporary file so the notebook does not depend on external files.

```
import shutil
import tempfile
from pathlib import Path

with tempfile.TemporaryDirectory() as folder:
    source = Path(folder) / "source.txt"
    destination = Path(folder) / "copy.txt"

    source.write_text("Python Standard Library")
    shutil.copyfile(source, destination)

    print(destination.read_text())
```

## 6. JSON
`json.dumps()` converts a Python object to JSON text, while `json.loads()` converts JSON text back to a Python object.

```
import json

data = {"name": "Ashish", "age": 28}

json_data = json.dumps(data)
print(json_data)
print(type(json_data))

python_data = json.loads(json_data)
print(python_data)
print(type(python_data))
```

## 7. CSV
The `csv` module is used to read and write comma-separated data.

```
import csv
from io import StringIO

data = StringIO()
writer = csv.writer(data)
writer.writerow(["name", "age", "marks"])
writer.writerow(["Ashish", 28, 89])

data.seek(0)
reader = csv.reader(data)

for row in reader:
    print(row)
```

## 8. Date and Time
`datetime` represents dates and times, while `timedelta` performs date/time calculations.

```
from datetime import datetime, timedelta

now = datetime.now()
yesterday = now - timedelta(days=1)

print("Now:", now)
print("Yesterday:", yesterday)
```

## 9. Time
The `time` module provides time-related functions. `sleep()` pauses program execution.

```
import time

start = time.time()
time.sleep(1)
end = time.time()

print(f"Elapsed time: {end - start:.2f} seconds")
```

## 10. Regular Expressions
The `re` module is used for pattern matching and text searching.

```
import re

text = "I am Ashish and I have 3 projects."
match = re.search(r"\d+", text)

if match:
    print("Number found:", match.group())
```

## Key Takeaways

- `math` → mathematical operations
- `random` → random values and selections
- `os` / `shutil` → file and system operations
- `json` / `csv` → data handling and serialization
- `datetime` / `time` → date and time operations
- `re` → pattern matching
- `array` → typed numeric arrays

> **Tip:** Prefer specific imports such as `from math import sqrt` over `from math import *` because they make code clearer and avoid namespace conflicts.
