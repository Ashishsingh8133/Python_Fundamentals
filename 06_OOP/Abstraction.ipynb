## Abstract Classes and Methods

Using Python's `abc` module to define abstract behaviour.

```python
from abc import ABC,abstractmethod
##creating abstract base class
class Vehicle(ABC):
    def drive(self):
        print("The car used for driving")
    
    @abstractmethod ##empty abstract method
    def start_engine(self):
        pass

##creating child class and inheriting abstract base class and implementing abstract method 
class car(Vehicle):  
    def start_engine(self):
        print(f"The car has started")

def operate_car(Vehicle): ##function to operate_car
    Vehicle.start_engine() ##acessing start enigne method
    Vehicle.drive() ##accessing drive method

##creating object 
car=car()
operate_car(car)
```
