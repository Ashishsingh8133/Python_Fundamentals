## Magic (Dunder) Methods

Demonstrating special methods that define how objects behave with Python operations and built-ins.

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
##create an object 
person=Person('Ashish',29)
print(person) ##when we execute this line then we get something like that this particular object created at this particular memory location because ther is 
#predefined magic method __str__ we can override this method to change the behaviour of our object
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    def __str__(self): ##Overriding the __str__ magic method
        return f"the age of {self.name} is {self.age}"

    def __repr__(self):
        return f"person(name={self.name},age={self.age})"


#creating object
person=Person('ashish',29)
print(person) ##after overriding the message will be different
print(repr(person))
    
```
