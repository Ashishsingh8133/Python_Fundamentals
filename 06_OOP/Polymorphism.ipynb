it allows a child class to provide a specific implementation of a method which is already present in it's parent class

## Polymorphism

Demonstrating how different objects can provide the same interface with different implementations.

```python
#Base class 1
class Animal:
     def speak(self): ##same method in parent class
         return "sound of animal"

#derived class 1
class Dog(Animal):
     def speak(self):  ## same method in child class 
         return "the dog speaks whoop! whoop!" ##here you can see that implementation is different


#Derived class 2
class cat(Animal):
     def speak(self): ##same method again
          return "the cat speaks Meow!" #different implementation 



##Function that demonstrate the polymorphism
def animal_speak(animal):
    print(animal.speak())
 
```

```python
dog=Dog() ## derived class 1 creating object
cat=cat() ## derived class 2 creating object
print(cat.speak())
print(dog.speak())
animal_speak(dog)
```

```python
##Base Class 1
class shape:
    def area(shape):
        return "the area of a figure"

#Derived class1
class rectangle(shape):
    def __init__(self,height,width):
        self.height=height
        self.width=width

    def area(self):
        return self.height*self.width
## Derived class 2
class circle(shape):
    def __init__(self,radius):
        self.radius=radius
    
    def area(self):
        return 3.14*self.radius*self.radius

##function that demonstarte polymorphism
def print_area(shape):
    print(f" the area is :{shape.area()}")

    
```

```python
rectangle=rectangle(4,5)
print_area(rectangle)
```

```python
circle=circle(4)
print_area(circle)
```

## Abstract Classes and Methods

Using Python's `abc` module to define abstract behaviour.

```python

###interface using abstractmethod

from abc import ABC,abstractmethod
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
     pass

class car(Vehicle):
    def start_engine(self):
        return "car engine is started"

class motorcycle(Vehicle):
    def start_engine(self):
        return "motorcycle engine is started"

def type_vehicle(Vehicle):
    print(Vehicle.start_engine())
```

```python
bullet=motorcycle()
type_vehicle(bullet)
```
