## Encapsulation

Demonstrating controlled access to class attributes and methods.

```python
class Person:
    def __init__(self,name,age):
        self.name=name ##public variables
        self.age=age   ##public variables
    

def get_detail(person):
    return person.name
```

```python
person=Person("ashish",29)
print(f"{person.name}")
print(f"{person.age}")
get_detail(person)
```

```python
dir(person) ## here we can see that vairable 'name' and 'age' is visible in the last of the list, it means they are public variables
```

```python
### create the private variables and try to access the value of it

class Person:
    def __init__(self,name,age,gender):
        self.__name=name ##private variable
        self.__age=age  ##private variable
        self.gender=gender ## public variable


def get_name(Person):
    return Person.__name ## trying to access __name which is private variable now     

def get_gender(Person):
    return Person.gender ##trying to access public variable
```

```python
person=Person("ashish",28,"male")
get_name(person) ##when we execute this code then as __name is private variable we will get an error we can't access the value of private variable
```

```python
get_gender(person) ## her we are able to access the public variable value bcz gender is public not private
```

```python
dir(person) ## here you can see that "name" and "age" will not appear in the list beacause they become private variables but gender is 
#public variable is appear in the last of the list
```

```python
##Protected Variables and accessing the value of protected variables
class Person:
    def __init__(self,name,age,gender):
        self._name=name
        self._age=age
        self.gender=gender
    

class Employee(Person): ##derived class for accesing the value of Protected variables
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)


##creating object and access value of protected variables
Human=Employee('ashish',29,'male')
print(Human._name)  ##here we are accesing the value of protected variables

##creating functions for direct access of the protected variable value from derived class "Employee"
def get_name(Employee):
    return Employee._name 

```

```python
get_name(Human) ##accessing value 
```

```python
##Accessing the private variable values
##Using getter and setter methods
class Person1:
    def __init__(self,name,age):
        self.__name=name
        self.__age=age
    
    def get_name(self):  ##getter method we are accessing value inside the class
     return self.__name

    def set_name(self,name): ##setter methods to update the value of private variable 
     self.__name=name

    def get_age(self): ##similar as the above one
     return self.__age

    def set_age(self,age):
        if age>0:
         self.__age=age
        else:
         print("invalid input")
     
        
     
#Creating object and output
marvel=Person1('ashish',29)
print(marvel.get_name())
print(marvel.get_age())
marvel.set_age(32)
print(marvel.get_age())
marvel.set_name('ashi')
print(marvel.get_name())
marvel.set_age(-5)
print(marvel.get_age())


```
