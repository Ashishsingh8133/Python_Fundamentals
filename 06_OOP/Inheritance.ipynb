SINGLE INHERITANCE 

## Inheritance Examples

Demonstrating how child classes inherit and extend parent-class behaviour.

```python
class car: ##parent class
    def __init__(self,windows,doors,enginetype): ##attributes of parent class
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):## instance methods of parent class
        print(f"The person is driving {self.enginetype} car")

```

```python
BMW=car(4,4,"Diesel")
BMW.drive()
```

## Using `super()`

Calling functionality from the parent class.

```python
## here we are creating a child class and inherit the parent class in child class
class Tesla(car):
    def __init__(self,windows,doors,enginetype,is_selfdriving): ## attributes of child class
        super().__init__(windows,doors,enginetype) ## attributes of parent class inherit using super()
        self.is_selfdriving=is_selfdriving ## adding the attribute in child class

    def selfdriving(self): #instance method of child class
        print(f"The Tesla is selfdriving car :{self.is_selfdriving}")
```

```python
spider6=Tesla(4,4,"electric",True)
spider6.selfdriving()
```

```python
spider6.drive() ## here it is inheriting the method of parent class and using the attribute value of child class which is "electric"
```

INHERIT MULTIPLE BASE CLASS FROM CHILD CLASS

```python
#BASE CLASS 1
class Animal:
    def __init__(self,name):
        self.name=name

    def speak(self):
        print(f"this is instance method 1")

#BASE CLASS 2
class pet:
    def __init__(self,owner):
        self.owner=owner
    
#DERIVED CLASS    
class dog(Animal,pet):
    def __init__(self,name,owner): ##derived class attributes
        Animal.__init__(self,name) ##base class1 inherited
        pet.__init__(self,owner) ##base class 2 inherited
    
    def speak(self):
        print(f"the animal {self.name} says whoop")
```

```python
pitbull=dog('rocky','ashish')
pitbull.speak()
print(f"Owner is :{pitbull.owner}")
```
