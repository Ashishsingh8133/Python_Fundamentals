```python
#instance variable and methods
class dog:
  ##constructor
  def __init__(self,name,age):
    self.name=name
    self.age=age

##creating object using class dog
dog1=dog('buddy',29) 
print(dog1)
```

```python
##instance methods 
class dog:
    ##instance variable
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
    ###instance methods
    def bark(self):
        print(f"{self.name} says whoop")
```

```python
dog1=dog('lucy',29)
print(dog1.name)
print(dog1.age)
dog1.bark()
```

```python
dog2=dog('tiger',30)
print(dog2.name)
print(dog2.age)
dog2.bark() ## calling instance methods using newly created objects
```

```python
#bank modelling
class BankAccount:
    ##constructor
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance
    
    #instance methods
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited and the new balance is {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            print(f"insufficient funds!")
        else:
            self.balance-=amount
            print(f"{amount} is withdraw,new balance is {self.balance}")
    def get_balance(self):
        print(f"{self.balance}")
```

```python
#create an account
account=BankAccount('Ashish',10000)
print(account.owner)
print(account.balance)

account.deposit(5000)
account.withdraw(2000)
account.withdraw(20000)
account.get_balance()
```
