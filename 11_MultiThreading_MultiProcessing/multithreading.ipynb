```python
###Single Thread 


import threading
import time

def print_numbers():
    for i in range(5):
        time.sleep(2) ##here each number will be printed after sleeping thread for 2 seconds it will take 2 seconds to print next number
        print(f"Numbers :{i}")

def print_Letters():
    for letter in "ashishsingh":
        time.sleep(2) ##here each character will be printed after sleeping for 2 second it will take 2 seconds to print next char
        #here the general concept is for using sleep is that there are some I/O operations 
        # happening behind so this thread will sleep for 2 seconds and leave the resource which will be used by another process 
        # and after 2 seconds it will continue the execution
        print(f"Letters are :{letter}")


start_time=time.time()
print("Process start time is :",start_time)

print_numbers()
print_Letters()
finish_time=time.time()-start_time
print("the process finish time is :",finish_time)

## Here only one thread is running line by line for these two function to improve the efficiency we can create multiple threads 
##here you can see that ther are 4 number needs to be executed and 11 char and 
# each number and each char is being executed after 2 seconds then for 16 execution it will take 32 seconds 
```

```python
##Here for these two functions we will create multiple threads so that there is no need to wait for line 
## by line execution we can execute these 2 function or processes parallel by creating multiple threads

import threading
import time

def print_numbers():
    for i in range(5):
        time.sleep(2) 
        print(f"Number:{i}")

def print_Letters():
    for letter in "abcde":
        time.sleep(2) 
        print(f"Letter:{letter}")


##creating multiple threads for these 2 processes
t1=threading.Thread(target=print_numbers)
t2=threading.Thread(target=print_Letters)


#process start time
t=time.time()
#print("Process start time is :",start_time)


##start the threads
t1.start()
t2.start()

#wait for the thread to complete
t1.join()
t2.join()

finish_time=time.time()-t
print("the process finish time is :",finish_time)

```
