## Function Decorators

Using decorators to extend or modify function behaviour.


## Applying a Decorator

Using `@decorator` syntax to apply a decorator to a function.


```python
@main_function ##this is decorator here by using @ we are calling the main function
def another_function(): #this function will go inside the main function as parameter and whatever will be present print along with main function
    print("it's very important to learn it")
```

## Wrapper Function

The wrapper function surrounds the original function and controls when it is executed.


```python
##example 2

def my_decorator(func):
    def wrapper():
        print("something is happening before the function is called")
        func()
        print("something is happening after the function is called")
    return wrapper ##here we are not returning the aub method to main function
```

```python
@my_decorator
def ashish():
    print("nothing is happening it's all a lie")
```

```python
ashish()
```

```python
##decorators with arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for _ in range(n):
                func(*args,**kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("say hello")
```

```python
say_hello()
```
