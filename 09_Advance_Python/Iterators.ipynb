## Creating an Iterator

Using Python's `iter()` function to obtain an iterator.


```python
##iterator
iterator=iter(fruits) ##iter is the keyword which is used to iterate over the given list or data-set
 #here it will not display the elements which are present in the fruits list
```

```python
##iterate through all the element
next(iterator) ##when we use next then it will display the first element of the list when we again execute this code then it will display the next element again and again
## till the last element in the list after that it will show the stop there are no elements in the list
```

```python
## when there will no elements in the list then we will get an exception "StopIteration" to handle this Exception
try:
    print(next(iterator))
except StopIteration:
    print("there are no elements in the list")
```

```python
##string iterator

my_string='ASHISH'
string_iterator=iter(my_string)
print(next(string_iterator))
print(next(string_iterator))
print(next(string_iterator))
print(next(string_iterator))
print(next(string_iterator))
print(next(string_iterator))
```

```python
number=[2,3,4,5,6]
number_iterator=iter(number)
```

```python
try:
    print(next(number_iterator))
except StopIteration:
    print("there are no numbers in the list")
```
