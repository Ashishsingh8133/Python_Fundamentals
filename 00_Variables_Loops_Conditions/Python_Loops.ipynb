Iterating over a sequence or collection using a `for` loop.

```python
for i in range (1,6):
    print(i)
```

## While Loops

Repeating a block of code while a condition remains `True`.

```python
#while loop
count=0
while count<5:
    print(count)
    count=count+1
```
