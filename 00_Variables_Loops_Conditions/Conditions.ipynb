## If Statements

Using `if` to execute code when a condition evaluates to `True`.

```python
age=int(input("Enter the value of num:"))
if age>=18:
    print("You are an adult")
else:
    print("You are a teenager")
```

```python
num=int(input("Enter the value of num:"))
if num>0:
    print("number is positive")
    if num%2==0:
        print("number is even")
    else:
        print("number is odd")
else:
    print("number is negative")            
    
```

## If, Elif and Else

Using conditional statements to execute different blocks of code based on conditions.

```python
a=int(input("Enter the value of a:"))
b=int(input("enter the value of b:"))
operation=str(input("Enter the operation u want to perform:"))
if operation=='add':
    result=a+b
elif operation=='sub':
    result=a-b
elif operation=='mul':
    result=a*b
elif operation=='div':
    if b!=0:
        result=a/b
    else:
        print("error")
else:
    
    print("Invalid input")
print("Result is:", result)                
    

      
```
