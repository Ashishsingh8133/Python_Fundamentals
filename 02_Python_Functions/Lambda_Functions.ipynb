# Lambda Functions

A lambda function is a small anonymous function written in a single expression. It is useful for short operations.

## 1. Lambda vs. Regular Function

```python
# Regular function
def add(a, b):
    return a + b

print(add(2, 3))

# Equivalent lambda function
add = lambda a, b: a + b
print(add(3, 4))
```

## 2. Lambda with a Condition

```python
# Regular function
def even(num):
    return num % 2 == 0

print(even(24))

# Equivalent lambda function
even = lambda num: num % 2 == 0
print(even(24))
```

## 3. Lambda for Calculations

```python
# Regular function
def square(number):
    return number ** 2

print(square(4))

# Equivalent lambda function
square = lambda number: number ** 2
print(square(5))
```

## 4. Lambda for Even/Odd Checking

```python
even_or_odd = lambda num: "Even" if num % 2 == 0 else "Odd"
print(even_or_odd(23))
```

### Key Takeaway

Use `lambda` for simple, short functions. For more complex logic, a regular `def` function is usually clearer.
