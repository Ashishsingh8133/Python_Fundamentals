# Filter Function

`filter()` selects elements from an iterable when the supplied function returns `True`.

## 1. Filter Even Numbers

```python
numbers = list(range(1, 13))

def is_even(num):
    return num % 2 == 0

print(list(filter(is_even, numbers)))

# Lambda version
print(list(filter(lambda x: x % 2 == 0, numbers)))
```

## 2. Filter Using Conditions

```python
numbers = list(range(1, 13))

# Keep numbers greater than 5
print(list(filter(lambda x: x > 5, numbers)))

# Keep even numbers greater than 5
print(list(filter(lambda x: x > 5 and x % 2 == 0, numbers)))
```

## 3. Filter a List of Dictionaries

```python
students = [
    {"name": "Ashish", "age": 28, "marks": 89},
    {"name": "Anmol", "age": 30, "marks": 92},
    {"name": "Anjali", "age": 26, "marks": 93}
]

# Keep students older than 25
def age_greater_than_25(student):
    return student["age"] > 25

print(list(filter(age_greater_than_25, students)))

# Lambda version
print(list(filter(lambda student: student["age"] > 25, students)))
```

### Key Takeaway

`filter()` is useful when you need to **select** items that satisfy a condition.
