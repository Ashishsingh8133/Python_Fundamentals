# Python Functions

This notebook covers the fundamentals of **Python functions**, including:

- Function syntax and docstrings
- Calling functions
- Functions with multiple parameters
- Default parameters
- Variable-length positional arguments (`*args`)
- Variable-length keyword arguments (`**kwargs`)
- Combining `*args` and `**kwargs`

The examples are intentionally simple so that each concept can be understood independently.


## 1. Function Syntax

A function is a reusable block of code defined using the `def` keyword. Parameters allow a function to receive input values, and `return` sends a result back to the caller.

```python
# Basic function syntax
def function_name(parameters):
    """Optional docstring describing what the function does."""

    # Function logic goes here.
    return statement
```

## 2. Function with a Single Parameter

The following function accepts one number and checks whether it is **even or odd** using the modulus (`%`) operator.

```python
def even_or_odd(num):
    """Print whether the supplied number is even or odd."""

    # A number is even when its remainder after division by 2 is 0.
    if num % 2 == 0:
        print(f"{num} is even")
    else:
        print(f"{num} is odd")
```

```python
# Call the function by passing an argument for the 'num' parameter.
even_or_odd(3)
```

## 3. Function with Multiple Parameters

A function can accept more than one parameter. Here, `a` and `b` are added together and the result is returned.

```python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b


# Pass two positional arguments to the function.
result = add(2, 4)

# Display the returned value.
print(result)
```

## 4. Default Parameters

A **default parameter** provides a value automatically when the caller does not supply that argument. The function below uses `Ashish` as the default name.

```python
def greet(name="Ashish"):
    """Print a welcome message using the supplied or default name."""
    print(f"Hey {name}, welcome to the parameters example!")


# No argument is supplied, so the default value is used.
greet()

# A supplied argument overrides the default value.
greet("Alex")
```

## 5. Variable-Length Positional Arguments (`*args`)

Use `*args` when a function should accept **any number of positional arguments**. Inside the function, `args` behaves like a tuple containing all supplied positional values.

```python
def print_numbers(*args):
    """Print each positional argument supplied to the function."""

    # Loop through every value collected by *args.
    for value in args:
        print(value)
```

```python
# The function can accept any number of positional arguments.
print_numbers(1, 2, 3, 4, 5, 6, 7, 8, "Ashish")
```

## 6. Variable-Length Keyword Arguments (`**kwargs`)

Use `**kwargs` when a function should accept **any number of keyword arguments**. Inside the function, `kwargs` behaves like a dictionary containing key-value pairs.

```python
def print_details(**kwargs):
    """Print each keyword argument as a key-value pair."""

    # .items() returns each key and its corresponding value.
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
# Keyword arguments are passed using the form key=value.
print_details(name="Ashish", age=17, marks=76)
```

## 7. Combining `*args` and `**kwargs`

A function can accept both variable-length positional and keyword arguments. Positional values are collected in `args`, while named values are collected in `kwargs`.

```python
def print_details_with_args(*args, **kwargs):
    """Print both positional and keyword arguments."""

    # Display values supplied as positional arguments.
    for value in args:
        print(f"Positional argument: {value}")

    # Display values supplied as keyword arguments.
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
# The first five values are positional arguments.
# name, age, and marks are keyword arguments.
print_details_with_args(
    1, 2, 3, 4, 5,
    name="Ashish",
    age=28,
    marks=89
)
```

## Key Takeaways

- `def` is used to define a function.
- Parameters allow functions to receive input.
- `return` sends a value back to the caller.
- Default parameters provide fallback values.
- `*args` collects variable-length positional arguments.
- `**kwargs` collects variable-length keyword arguments.
- `*args` is stored as a tuple, while `**kwargs` is stored as a dictionary.
