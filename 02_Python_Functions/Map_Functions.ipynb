# Map Function

`map()` applies a function to every item in an iterable and returns the transformed results.

## 1. Square Every Number

```python
numbers = [2, 3, 4, 5, 6]

def square(number):
    return number ** 2

print(list(map(square, numbers)))

# Lambda version
print(list(map(lambda x: x ** 2, numbers)))
```

## 2. Add Two Lists

```python
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]

added_list = list(map(lambda x, y: x + y, numbers1, numbers2))
print(added_list)
```

## 3. Convert Strings to Integers

```python
string_numbers = ["1", "2", "3", "4"]

integer_numbers = list(map(int, string_numbers))
print(integer_numbers)
```

## 4. Convert Strings to Uppercase

```python
words = ["apple", "banana", "grapes"]

upper_words = list(map(str.upper, words))
print(upper_words)
```

## 5. Map over a List of Dictionaries

```python
students = [
    {"name": "Ashish", "age": 28, "marks": 89},
    {"name": "Anmol", "age": 30, "marks": 91},
    {"name": "Ankit", "age": 32, "marks": 92}
]

# Extract the name from each dictionary
def get_name(student):
    return student["name"]

print(list(map(get_name, students)))

# Lambda version
print(list(map(lambda student: student["name"], students)))
```

### Key Takeaway

`map()` is useful when you need to **transform** every item in an iterable.
