## Custom Exceptions

Defining a custom exception class for application-specific errors.

```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

## Raising Exceptions

Using `raise` to deliberately trigger an exception when a condition is not valid.

```python
year=int(input("enter the age"))
age=2024-year
                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                                          
try:
    if age<=30 and age>=20:
        print("You are eligible to fill this form")
    else:
       raise dobException
except dobException:
    print("sorry your age should be greator than 20 or less than 30")


```

## Key Takeaways

- Understand how exceptions are handled in Python.
- Use `try`, `except`, `else`, and `finally` appropriately.
- Use `raise` when an exception needs to be triggered explicitly.
- Custom exceptions can make error handling clearer for specific applications.
