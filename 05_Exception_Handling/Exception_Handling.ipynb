## Try–Except

Handling errors safely using `try` and `except`.

```python
##Exception try,except Block
try:
    a=b
except:
    print("the value of b Variable has not defined")
```

## Custom Exceptions

Defining a custom exception class for application-specific errors.

```python
try:
    a=b
except NameError as bh: #this NameError is class which is going to handle this kind of exception
    print(bh)
```

```python
result=1/0 ##Zero Division Error
print(result)
```

```python
try:
    result=1/0
except ZeroDivisionError  as gv: #zerodivision error is a class which is going to handle this exception 
    print(gv)
```

```python
try:
    result=1/2
    a=b
except ZeroDivisionError  as gv: 
    print(gv)
except Exception as ex: #this is the base class which can handle any type of exception
    print(ex)
    print("main exception got caught here")
```

```python
try:
    num=int(input("Enter the value of num"))
    result=10/num
except ValueError:
    print("enter valid value")
except ZeroDivisionError:
    print("Enter the Denominator Value Greator Than Zero")
except Exception as ex: ##the base class which is going to handle any type of excceptions
    print(ex)
```

## Else Block

Using `else` for code that should run when no exception occurs.

```python
#try,except and else
try:
    num=int(input("Enter the Value of num"))
    result=10/num
except ValueError:
    print("Enter the valid value")
except ZeroDivisionError:
    print("Enter the denominator value greator than 0")
except Exception as ex1:
    print(ex1)
else:
    print(f"the result is:{result}")
```

```python
try:
    num=int(input("Enter the Value of num"))
    result=10/num
except ValueError:
    print("Enter the valid value")
except ZeroDivisionError:
    print("Enter the denominator value greator than 0")
except Exception as ex1:
    print(ex1)
else:
    print(f"the result is:{result}")
finally:
    print("Execution completed")
```

## Finally Block

Using `finally` for code that should execute whether an exception occurs or not.

```python
#exception handling and file handling

try:
    file=open('example.txt',mode='r')
    content=file.read()
    print(content)
    a=b

except FileNotFoundError:
    print("file does not found")
except Exception as ex1:
    print(ex1)
finally:
    if 'file' in locals() or not file.closed():
        file.close()
        print("file is closed")
```

## Key Takeaways

- Understand how exceptions are handled in Python.
- Use `try`, `except`, `else`, and `finally` appropriately.
- Use `raise` when an exception needs to be triggered explicitly.
- Custom exceptions can make error handling clearer for specific applications.
