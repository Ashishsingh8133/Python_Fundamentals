###Reference Counting 

```python
##refrence counting 

import sys
a=[]

##(we are getting 2 as a refrence count because one refrence from 'a' and one from getrefcount())
print(sys.getrefcount(a))
```

```python
b=a 

##here we get 3 as reference count for b because a has already 2 counts and when a assigned to b then 1 count added for b total refrence count is 3
print(sys.getrefcount(b))
```

```python
del b
print(sys.getrefcount(b)) ##since 'b' has deleted so it has no refrence count 
```

```python
###after deleting b when we count the refrence count of a then we get 2 because a has 2 reference count 
print(sys.getrefcount(a))
```

##GARBAGE COLLECTOR

```python
##to enable Garbage collector
import gc
gc.enable() #here we enable the garbage collector
```

```python
gc.collect() ##there are currently 600 objects in the system and they are refering to each other to prevent their refrence count reach to zero
```

```python
gc.disable() ##here we disable the garbage collector
```

```python
## get garbage collection stats
print(gc.get_stats())
```

```python
##get unreachable objects
print(gc.garbage) ##it will print unreachable objects
```

```python
##Example of best Memory Practices
##An example what will happen when circular refrences not managed well, there will be memory leak
import gc

class MyObject:
    def __init__(self,name):
        self.name=name
        print(f"Object {self.name} is created")

    def __del__(self):
        print(f"Object {self.name} deleted")


# create objects
obj1=MyObject("obj1")
obj2=MyObject("obj2")

#create circular refrences
obj1.ref=obj2
obj2.ref=obj1

del obj1 ##here when we delete the object both like obj1 and obj2 and i will not call that del function then because of circular refrence there will be memory leak
del obj2

##manually trigger the garbage the garbage collection
gc.collect() ##when we manually call this then only obj1 and obj2 will be deleted
 
```

```python
##Print collected objects
print(f"Garbage collected objects: {gc.garbage}")
```

```python
##Generators for memory efficiency
## Generators allows you to produce one item at a time on the memory to use memory efficiently
def generate_number(n):
    for i in range(n):
        yield i

for num in generate_number(10):
    print(num)
    
```

```python
##profiling memory usage with Tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()

    snapshot=tracemalloc.take_snapshot()
    top_stats=snapshot.statistics('lineno')

    print("[ Top 10 ]")
    for stat in top_stats[:5]:
        print(stat)
```

```python
main()
```
